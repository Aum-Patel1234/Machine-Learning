```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("data/train.csv")
test_df = pd.read_csv("data/test.csv")
```

```python
for obj_col in  df.select_dtypes(include='object'):
    print(obj_col, " - ", df[obj_col].unique())
```

```python
df1 = df.drop(["id", "Irrigation_Need"], axis=1)
```

```python
obj_cols = df1.select_dtypes(include='object').columns
obj_cols
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
col_transformer = ColumnTransformer(
    transformers=[
        ('onehot', OneHotEncoder(drop='first', sparse_output=False), obj_cols)
    ],
    remainder='passthrough'
)
```

```python
df1_transformed = col_transformer.fit_transform(df1)
```

```python
new_cols = col_transformer.get_feature_names_out()
new_cols
```

```python
df2 = pd.DataFrame(df1_transformed, columns=new_cols)
```

```python
assert(len(df2.select_dtypes(include='object').columns) == 0)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
```

```python
models = {
    "logistic_regression": LogisticRegression(),
    "decision_tree": DecisionTreeClassifier(),
    "xgboost": XGBClassifier(),
    "catboost": CatBoostClassifier(verbose=0)
}
```

```python
y = df['Irrigation_Need']
```

```python
mapping = {'Low': 0, 'Medium': 1, 'High': 2}

y_train = y.map(mapping)
```

```python
X_train, X_test, y_train, y_test = train_test_split(df2, y_train, test_size=0.2, random_state=42)
```

```python
from sklearn.metrics import balanced_accuracy_score
```

```python
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    print(name, " - ", model.score(X_test, y_test))
    print("balanced_accuracy_score =", balanced_accuracy_score(y_test, y_pred), '\n')
```

```python
test_df1 = test_df.drop('id', axis=1)
```

```python
test_transformed = col_transformer.transform(test_df1)
```

```python
test_transformed_df = pd.DataFrame(
    test_transformed,
    columns=col_transformer.get_feature_names_out(),
    index=test_df1.index
)
```

```python
pred = models['catboost'].predict(test_transformed_df)
```

```python
reverse_mapping = {v: k for k, v in mapping.items()}
```

```python
y_pred_labels = pd.Series(pred.ravel()).map(reverse_mapping)
```

```python
sub_df = pd.DataFrame({
    "id": test_df["id"],
    "Irrigation_Need": y_pred_labels
})
```

```python
sub_df.to_csv("submission.csv", index=False)
```
