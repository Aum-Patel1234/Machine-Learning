```python
def target_encode(train_df, test_df, cols, target):
    train_df = train_df.copy()
    test_df = test_df.copy()

    global_mean = train_df[target].mean()

    for col in cols:
        # ensure consistent dtype
        train_df[col] = train_df[col].astype(str)
        test_df[col] = test_df[col].astype(str)

        mapping = train_df.groupby(col)[target].mean()

        # apply mapping
        train_df[col + '_enc'] = train_df[col].map(mapping)
        test_df[col + '_enc'] = test_df[col].map(mapping)

        # handle unseen categories
        test_df[col + '_enc'] = test_df[col + '_enc'].fillna(global_mean)

    return train_df, test_df
```

```python
cols = ['Driver', 'Race']
target = 'PitNextLap'

df3, test_df = target_encode(df2, test_df, cols, target)
```
